```python
import pandas as pd
import kagglehub

# 1. Descargamos el dataset forzando a ignorar el caché corrupto local
ruta_carpeta = kagglehub.dataset_download(
    "iammustafatz/diabetes-prediction-dataset",
    force_download=True
)

# 2. Construimos la ruta exacta al archivo CSV
ruta_csv = f"{ruta_carpeta}/diabetes_prediction_dataset.csv"

# 3. Leemos el archivo directamente con pandas
df = pd.read_csv(ruta_csv)

print("--- Información General ---")
df.info()
```

```python

# 2. El estado mecánico: Te da un resumen estadístico de las columnas numéricas.
# Aquí puedes ver el promedio, el valor mínimo y el máximo. ¡Ideal para detectar anomalías!
# (Por ejemplo, si el nivel de glucosa mínimo es 0, sabemos que hay un error de captura)
print("\n--- Resumen Estadístico ---")
print(df.describe())

# 3. La prueba de fugas: Te muestra exactamente cuántos valores faltantes hay por columna
print("\n--- Valores Nulos por Columna ---")
print(df.isnull().sum())
```
